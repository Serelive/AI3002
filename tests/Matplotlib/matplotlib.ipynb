```python
import matplotlib.pyplot as plt

import numpy as np
import pandas as pd
```

```python
x = np.linspace(0, 10, 101)
y = np.sin(x)

fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_title("Sine Function")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
plt.show()
```

```python
x = np.linspace(0, 10, 100)

fig, ax = plt.subplots()

ax.plot(x, np.sin(x), label="sin(x)")
ax.plot(x, np.cos(x), label="cos(x)", linestyle="--")

ax.set_xlabel("x")
ax.set_ylabel("Value")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

```python
rng = np.random.default_rng(42)

x = np.linspace(1,10,50)
y = 2 * x + 1 + rng.normal(0, 2, size=50)

fig, ax = plt.subplots()

ax.scatter(x, y, label="Data")
ax.plot(x, 2 * x + 1, label="True")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()

plt.show()
```
